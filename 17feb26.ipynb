You are building a small student data management tool. Each student is identified by a roll number stored in a Python list. Write a Python program that performs the following operations on the student roll numbers list: Reads the initial list of roll numbers. Extends the list with another batch of roll numbers. Appends a new admission roll number. Inserts a student roll number at a specific index. Counts how many times a roll number R appears in the list. Finds the index of a roll number S. Removes an incorrect roll number from the list. Sorts the final list of roll numbers. Prints the final sorted list, count of R, and index of S. Input Format Integer N — number of initial roll numbers N space-separated roll numbers Integer M — number of roll numbers to extend M space-separated roll numbers to extend New admission roll number to append A roll number to insert Index to insert at (0-based) R — roll number to count S — roll number to find index of Incorrect roll number to remove Constraints NA Output Format First line: final sorted roll list, space-separated Second line: count of R Third line: index of S Sample Input 0 5 10 20 30 20 40 3 50 60 70 80 25 1 20 30 40 Sample Output 0 10 20 20 25 30 50 60 70 80 2 3 Sample Input 1 5 101 102 103 104 105 3 106 107 108 109 110 2 102 105 103 Sample Output 1 101 102 104 105 106 107 108 109 110 1 5

```python
# Step 1: Read initial list
n = int(input())
lst = list(map(int, input().split()))

# Step 2: Extend list
m = int(input())
lst.extend(list(map(int, input().split())))

# Step 3: Append new roll number
append_val = int(input())
lst.append(append_val)

# Step 4: Insert at index
insert_val = int(input())
index = int(input())
lst.insert(index, insert_val)

# Step 5: Count occurrences of R
R = int(input())
count_R = lst.count(R)

# Step 6: Find index of S
S = int(input())
index_S = lst.index(S)

# Step 7: Remove incorrect roll number
remove_val = int(input())
lst.remove(remove_val)

# Step 8: Sort list
lst.sort()

# Output
print(*lst)
print(count_R)
print(index_S)
```
